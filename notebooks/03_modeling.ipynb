# 03 - Model Training

Trains Logistic Regression, Random Forest, and XGBoost on the SMOTE-balanced training set with 5-fold cross-validation.

```python
import os
import sys

ROOT = os.path.abspath(os.path.join(os.getcwd(), '..')) if os.path.basename(os.getcwd()) == 'notebooks' else os.getcwd()
sys.path.insert(0, ROOT)

from src.train import load_train_data, apply_smote, train_all_models
```

## 1. Load processed training data and apply SMOTE

```python
X, y = load_train_data()
print(f'Loaded train: {X.shape}, fraud rate = {y.mean():.4f}')
X_sm, y_sm = apply_smote(X, y)
print(f'After SMOTE - Fraud: {int((y_sm==1).sum())}, Legit: {int((y_sm==0).sum())}')
```

## 2. Train all models (saves pickles + best_model.pkl)

```python
results = train_all_models(X_sm, y_sm)
```

## 3. Cross-validation summary

```python
import pandas as pd
summary = pd.DataFrame([{'Model': k, 'CV_ROC_AUC': v['cv_auc']} for k, v in results.items()])
summary.sort_values('CV_ROC_AUC', ascending=False)
```

```python
models_dir = os.path.join(ROOT, 'models')
print(sorted(f for f in os.listdir(models_dir) if f.endswith('.pkl')))
```
